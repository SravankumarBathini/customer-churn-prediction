```python
# CELL 1: ENTERPRISE ENVIRONMENT & PACKAGE VALIDATION LAYER
import polars as pl
import numpy as np
import xgboost as xgb
import imblearn

print("✅ ENVIRONMENT INTEGRITY CHECK: PASSED")
print(f"   Polars Processing Engine Engine Layer Version : {pl.__version__}")
print(f"   XGBoost Sequential Gradient Boosting Version : {xgb.__version__}")
print(f"   Imbalanced-Learn Sampling Layer Version      : {imblearn.__version__}")
```

```python
# CELL 2: HIGH-SPEED RUST-BACKED DATA SYNTHESIS & IMBALANCE LAYER
import polars as pl
import numpy as np

print("📡 Synthetic Engine Active: Fabricating 100,000 enterprise subscriber profiles...")

# Set seed for mathematical reproducibility across all engineering runs
np.random.seed(42)
total_records = 100000
churn_rate = 0.05
minority_count = int(total_records * churn_rate)      # 5,000 Churn events
majority_count = total_records - minority_count       # 95,000 Stay events

# 1. Synthesize Features for Class 0 (The Loyal Majority)
majority_service_calls = np.random.poisson(lam=1.5, size=majority_count)
majority_charges = np.random.normal(loc=65.0, scale=15.0, size=majority_count)
majority_contract = np.random.choice([12, 24], size=majority_count, p=[0.4, 0.6])
majority_labels = np.zeros(majority_count, dtype=np.int32)

# 2. Synthesize Features for Class 1 (The Volatile Churning Minority)
# Churning customers characteristically show higher service friction and inflated pricing bills
minority_service_calls = np.random.poisson(lam=4.2, size=minority_count)
minority_charges = np.random.normal(loc=95.0, scale=20.0, size=minority_count)
minority_contract = np.random.choice([0, 12], size=minority_count, p=[0.8, 0.2])
minority_labels = np.ones(minority_count, dtype=np.int32)

# 3. Concatenate arrays and compile into an optimized Polars DataFrame
df_telecom = pl.DataFrame({
    "customer_service_calls": np.concatenate([majority_service_calls, minority_service_calls]),
    "monthly_charges_usd": np.concatenate([majority_charges, minority_charges]),
    "contract_duration_months": np.concatenate([majority_contract, minority_contract]),
    "subscriber_churn_status": np.concatenate([majority_labels, minority_labels])
})

# Shuffle the dataframe to simulate raw, unstructured warehouse ingestion pools
df_telecom = df_telecom.sample(fraction=1.0, shuffle=True, seed=42)

# 4. Execute a Corporate Audit Report on Class Distributions
distribution_report = df_telecom.group_by("subscriber_churn_status").count()
print("\n📋 ================================================")
print("     ENTERPRISE DATAWAREHOUSE INGESTION PROFILE")
print("==================================================")
for row in distribution_report.iter_rows(named=True):
    status = "⚠️ CHURN (Minority)" if row["subscriber_churn_status"] == 1 else "✅ STAY (Majority)"
    percentage = (row["count"] / total_records) * 100
    print(f"  🔹 Account Status: {status:<18} | Allocation: {row['count']:>6} accounts ({percentage:.2f}%)")
print("==================================================")
```

```python
# CELL 3: DECOUPLED MATRIX EXTRACTION & STRATIFIED PARTITION LAYER
from sklearn.model_selection import train_test_split

print("🔌 Extracting telecom subscriber matrices from ingestion layers...")

# 1. Isolate historical clue matrices (X) from the true target answer (y)
X_telecom = df_telecom.select(["customer_service_calls", "monthly_charges_usd", "contract_duration_months"])
y_telecom = df_telecom.select("subscriber_churn_status")

# 2. Execute a strict 80% Training / 20% Testing Stratified Split
# By declaring 'stratify=y_telecom.to_pandas()', we force the split to preserve our 5% minority proportions
X_train_raw, X_test, y_train_raw, y_test = train_test_split(
    X_telecom.to_pandas(),
    y_telecom.to_pandas().values.ravel(),
    test_size=0.20,
    random_state=42,
    stratify=y_telecom.to_pandas().values.ravel()
)

print("\n✂️ Bounded Stratified Partitioning Complete! Metric Allocations:")
print(f" 📂 X_train_raw (Homework Clues) : {X_train_raw.shape} rows/columns")
print(f" 📂 X_test      (Final Exam Clues): {X_test.shape} rows/columns")

# 3. Verify and audit the mathematical class balance inside each matrix split
train_churn_pct = (np.sum(y_train_raw == 1) / len(y_train_raw)) * 100
test_churn_pct = (np.sum(y_test == 1) / len(y_test)) * 100

print("\n📋 ================================================")
print("      STRATIFICATION CLASS INTEGRITY REPORT")
print("==================================================")
print(f"  🔹 Training Set Minority Baseline : {np.sum(y_train_raw == 1):>4} accounts ({train_churn_pct:.2f}%)")
print(f"  🔹 Testing Set Minority Baseline  : {np.sum(y_test == 1):>4} accounts ({test_churn_pct:.2f}%)")
print("==================================================")
print("✅ SUCCESS: Proportional ratio balance preserved across both data horizons.")
```

```python
# CELL 4: SYNTHETIC MINORITY OVER-SAMPLING (SMOTE) LAYER
from imblearn.over_sampling import SMOTE
from collections import Counter
import numpy as np

print("🧬 Initializing Synthetic Minority Over-sampling Technique (SMOTE)...")

# 1. Instantiate the SMOTE engine core
# We set random_state to ensure the synthetic lines match perfectly across runs
smote_engine = SMOTE(random_state=42)

# 2. Fit the resampling layer EXCLUSIVELY to our training matrix
print("🏋️ Extracting nearest neighbor paths and synthesizing new subscriber records...")
X_train_balanced, y_train_balanced = smote_engine.fit_resample(X_train_raw, y_train_raw)

print("\n🚀 SMOTE Resampling Complete! Balanced Matrix Allocations:")
print(f" 📂 X_train_balanced (Balanced Homework Clues) : {X_train_balanced.shape} rows/columns")
print(f" 📂 y_train_balanced (Balanced Homework Keys)  : {y_train_balanced.shape} rows")

# 3. Execute a Corporate Audit Report on the fresh distributions
balanced_counts = Counter(y_train_balanced)
total_balanced = len(y_train_balanced)

print("\n📋 ================================================")
print("        BALANCED TRAINING DATA ARCHITECTURE REPORT")
print("==================================================")
print(f"  🔹 Total Post-SMOTE Training Records : {total_balanced:,} accounts")
print(f"  🔹 Class 0 (Loyal Majority Account) : {balanced_counts[0]:,} rows ({ (balanced_counts[0]/total_balanced)*100 :.2f}%)")
print(f"  🔹 Class 1 (Synthetic Churn Account) : {balanced_counts[1]:,} rows ({ (balanced_counts[1]/total_balanced)*100 :.2f}%)")
print("==================================================")
print("✅ SUCCESS: Training matrix has achieved a perfect 50/50 equilibrium.")
```

```python
# CELL 5: SEQUENTIAL ENSEMBLE GRADIENT BOOSTING LAYER
import xgboost as xgb
import numpy as np

print("🏎️ Initializing High-Performance XGBoost Sequential Ensemble Engine...")

# 1. Configure the enterprise hyperparameters for the tree ensemble
# We use 'binary:logistic' to output raw probabilities, and use logloss to guide optimization
telecom_ensemble = xgb.XGBClassifier(
    n_estimators=150,           # Deploy a committee of exactly 150 sequential trees
    max_depth=4,                # Bounded tree depth to force generalized pattern learning
    learning_rate=0.05,         # Scaling step-size shrinkage to prevent training overshoot
    objective="binary:logistic",
    eval_metric="logloss",      # Guide the sequential optimization using LogLoss penalty curves
    random_state=42,
    tree_method="hist"          # Leverage optimized histogram binning for high-speed computation
)

# 2. Train the ensemble committee exclusively on our Balanced Homework Matrix
print("🏋️ Exposed to post-SMOTE training matrix. Growing tree sequence...")
telecom_ensemble.fit(X_train_balanced, y_train_balanced)
print("🌲 Model training complete! 150 sequential weak learners frozen successfully.")

# 3. Simulate the Final Exam: Extract RAW probabilities across our untouched test records
print("📡 Dispatching probability predictions across the 20,000 unseen test records...")
unseen_probabilities = telecom_ensemble.predict_proba(X_test)[:, 1]

# 4. Preview the first 5 customer comparisons to check calibration manually
print("\n🔍 Live Diagnostic Sample Check (True Status vs. Model Churn Confidence):")
for i in range(5):
    true_label = "⚠️ CHURN" if y_test[i] == 1 else "✅ STAY"
    print(f"   Subscriber {i+1}: True Status = {true_label:<8} ➔ Model Forecast Risk = {unseen_probabilities[i]*100:.2f}%")
```

```python
# CELL 6: ENTERPRISE MODEL EVALUATION & CONFUSION MATRIX LAYER
from sklearn.metrics import confusion_matrix, classification_report
import numpy as np

print("📊 Dissecting Sequential Ensemble Performance Across Unseen Test Sets...")

# 1. Convert our fluid risk probabilities into hard binary decisions using a strict 50% gate
hard_predictions = (unseen_probabilities >= 0.50).astype(np.int32)

# 2. Extract the exact count locations for our 4 critical quadrants
tn, fp, fn, tp = confusion_matrix(y_test, hard_predictions).ravel()

# 3. Print out the formal Corporate Evaluation Report
print("\n📋 ===============================================")
print("     ENTERPRISE MODEL CONFUSION MATRIX REPORT")
print("===============================================")
print(f"   True Negatives  (Correctly Predicted Stays) : {tn:,}")
print(f"   False Positives (False Alarms / Wasted Ads) : {fp:,}")
print(f"   False Negatives (Missed Churns / Revenue Leak): {fn:,}")
print(f"   True Positives  (Successfully Caught Churns): {tp:,}")
print("===============================================")

# 4. Generate the Precision, Recall, and F1-Score breakdown for the minority class (1)
report_dict = classification_report(y_test, hard_predictions, output_dict=True)
precision_score = report_dict["1"]["precision"]
recall_score = report_dict["1"]["recall"]

print("\n💡 Core Architectural Grading Performance Metrics:")
print(f"   🔹 Model Precision Score (False Alarm Cleanliness): {precision_score*100:.2f}%")
print(f"   🔹 Model Recall Score    (Churn Catch Rate)        : {recall_score*100:.2f}%")

if recall_score < 0.85:
    print("\n⚠️ WARNING: Revenue leak detected. The model's blind spot is letting too many churners slip past.")
else:
    print("\n✅ SUCCESS: The synthetic SMOTE training layer has successfully eliminated structural bias.")
    print("   The ensemble captures over 85% of high-risk customer turnover windows predictably.")
```

```python
# CELL 7: FEATURE IMPORTANCE TOPOGRAPHY LAYER
import pandas as pd
import matplotlib.pyplot as plt

print("📊 Querying structural layers of the sequential tree committee...")

# 1. Extract raw gain-based feature weights straight from the trained XGBoost engine
raw_importances = telecom_ensemble.feature_importances_
feature_names = X_train_raw.columns

# 2. Map weights to their respective column fields and sort into a clear hierarchy
importance_bridge = pd.DataFrame({
    "Feature_Asset": feature_names,
    "Relative_Weight_Score": raw_importances
}).sort_values(by="Relative_Weight_Score", ascending=False)

# 3. Print out the formal Corporate Risk Ingestion Report
print("\n📋 ===============================================")
print("       ENTERPRISE RISK FACTOR HIERARCHY REPORT")
print("===============================================")
for rank, row in enumerate(importance_bridge.itertuples(), 1):
    percentage = row.Relative_Weight_Score * 100
    print(f"  🏅 Rank {rank}: {row.Feature_Asset:<25} | Structural Impact: {percentage:.2f}%")
print("===============================================")
print("✅ SUCCESS: Model inner-logic fully deciphered and exported for stakeholder review.")
```
